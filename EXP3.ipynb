```python
print("CONVERSATIONAL AI CHATBOT")
print("-" * 35)

print("Chatbot: Hello! Welcome to the AI Chatbot.")
print("Chatbot: Type 'exit' to end the conversation.")

while True:

    user = input("\nYou: ")

    if user.lower() == "exit":
        print("Chatbot: Thank you! Goodbye.")
        break

    elif "hello" in user.lower() or "hi" in user.lower():
        print("Chatbot: Hello! How can I help you?")

    elif "name" in user.lower():
        print("Chatbot: I am a simple AI chatbot.")

    elif "ai" in user.lower():
        print("Chatbot: Artificial Intelligence enables")
        print("machines to perform intelligent tasks.")

    elif "python" in user.lower():
        print("Chatbot: Python is a popular programming language.")

    else:
        print("Chatbot: I received your message.")
        print("Chatbot: Please ask another question.")
```
