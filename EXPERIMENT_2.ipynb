```python
!pip install -q transformers torch

from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="gpt2"
)

# Prompt 1 - Content Generation
prompt1 = """
Write a short paragraph about the importance of Artificial Intelligence
in education.
"""

result1 = generator(
    prompt1,
    max_new_tokens=100,
    do_sample=True
)

print("===== CONTENT GENERATION =====")
print(result1[0]["generated_text"])


# Prompt 2 - Task Automation
prompt2 = """
Task: Create a simple daily study plan for a computer science student.
"""

result2 = generator(
    prompt2,
    max_new_tokens=100,
    do_sample=True
)

print("\n===== TASK AUTOMATION =====")
print(result2[0]["generated_text"])
```
